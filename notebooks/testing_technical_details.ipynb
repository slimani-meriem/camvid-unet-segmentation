```python
R = 128
G = 64
B = 192

print(f" R = {R} : binaryR = {bin(R)}")
print(f" G = {G} : binaryG = {bin(G)}")
print(f" B = {B} : binaryB = {bin(B)}")

encoded = (R<<16) | (G<<8) | B
print(f" encoded = {encoded} : binaryEncoded = {bin(encoded)}")

encoded = (R<<16) + (G<<8) + B
print(f" encoded = {encoded} : binaryEncoded = {bin(encoded)}")
```
